## Importing:

```python
import numpy as np

X = np.array([
    [500],
    [700],
    [900],
    [1100],
    [1300],
    [1500]
])

y = np.array([
    100,
    140,
    180,
    220,
    260,
    300
])
```

## Import Tensor flow:

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

## Create a model:

```python
model = Sequential([
    Dense(10,activation="relu",input_shape=(1,)),
    Dense(1)
    ])
```

## Compile the model

```python
model.compile(
    optimizer="adam",
    loss = "mse"
)
```

## Train the network:

```python
model.fit(X,y,epochs=100,verbose=1,batch_size=32)
```

```python
prediction = model.predict(
    np.array([[1700]])
)

print(prediction)
```
