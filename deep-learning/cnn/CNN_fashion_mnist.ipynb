```python
input_layer = tf.keras.layers.Input(shape=(28, 28, 1))

model = tf.keras.Sequential([
    input_layer,

    layers.Conv2D(
        filters=10,
        kernel_size=(3, 3),
        activation='relu'
    ),

    layers.Conv2D(
        filters=10,
        kernel_size=(3, 3),
        activation='relu'
    ),

    layers.MaxPooling2D(),

    layers.Conv2D(
        filters=10,
        kernel_size=(3, 3),
        activation='relu'
    ),

    layers.Conv2D(
        filters=10,
        kernel_size=(3, 3),
        activation='relu'
    ),

    layers.MaxPooling2D(),

    layers.Flatten(),

    layers.Dense(10, activation='softmax')
])
```
