# Date Time / Time Delta in python:

```python
from datetime import date, timedelta
```

## Get today's date:

```python
today = date.today()
```

## Convert today's date into text:

```python
d1 = today.strftime("%Y-%m-%d")
```

## Get the date one year ago:

```python
d2 = (today - timedelta(days=365)).strftime("%Y-%m-%d")
```

```python
start_date = d2
end_date = d1
```

```python
tickerSymbol = 'META'

tickerData = yf.Ticker(tickerSymbol)
```

```python
df =tickerData.history(start=start_date, end=end_date)
```

```python
tickerData.info
```

```python
df.to_csv('META.csv')
```
