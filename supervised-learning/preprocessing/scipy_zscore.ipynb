## Z- Score:

```python
import pandas as pd

df = pd.DataFrame({
    "Marks": [40, 50, 60, 70, 80]
})
```

## calculate mean:

```python
mean = df["Marks"].mean()
```

### Calculate Z-score:

```python
df["z-score"] = (df["Marks"]-mean)/df["Marks"].std()
```

```python
print(df)
```

```python
plt.figure(figsize=(7,4))

plt.scatter(df["Marks"], df["z-score"], s=100)

plt.axhline(0, color="red", linestyle="--", label="Mean (Z=0)")

plt.xlabel("Marks")
plt.ylabel("Z-score")
plt.title("Marks vs Z-score")
plt.legend()

plt.show()
```
