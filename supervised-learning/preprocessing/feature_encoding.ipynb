# Import Libraries:


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
```

## Label Encoder:

```python
le = LabelEncoder()
gender = ["Male","Female","Trans"]
encoded = le.fit_transform(gender)
print(encoded)
```

## One Hot Encoding:

```python
from sklearn.preprocessing import OneHotEncoder
df = pd.DataFrame({
    "City": ["Lahore", "Karachi", "Lahore"]
})
ohe = OneHotEncoder(sparse_output=False)
encoded = ohe.fit_transform(df[["City"]])
print(encoded)
```

## Ordinal Encoder:

```python
from sklearn.preprocessing import OrdinalEncoder
df = pd.DataFrame({
    "Size":["Small","Medium","Large","Small","Medium"]
})
oe = OrdinalEncoder(categories=[["Small","Medium","Large"]])
encoded = oe.fit_transform(df[["Size"]])
```

## Feature Encoding in pandas:

```python
import pandas as pd

df = pd.DataFrame({
    "City": ["Lahore", "Karachi", "Islamabad", "Lahore"]
})

print(df)
encoded_df = pd.get_dummies(df)
```

* # Discretization:

```python
from sklearn.preprocessing import KBinsDiscretizer
df2 = pd.DataFrame({
    "Age":[22,25,30,35,40,45,50]
})
print(df2)
```

```python
kbd = KBinsDiscretizer(n_bins=3,encode="ordinal",strategy="uniform")
df2["Age_binned"] = kbd.fit_transform(df2[["Age"]])
```
