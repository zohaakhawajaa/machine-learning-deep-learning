# Import Libraries:

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, MaxAbsScaler,RobustScaler
```

## Import the scaler:

>Dataset:


```python
import pandas as pd

df = pd.DataFrame({
    "Age": [18, 22, 25, 30, 35, 40],
    "Salary": [25000, 35000, 45000, 60000, 80000, 100000],
    "Experience": [0, 1, 3, 5, 8, 12]
})

print(df)
```

## Standard Scaler:

```python
scaler = StandardScaler()
scaled = scaler.fit_transform(df)
print(pd.DataFrame(scaled,columns=df.columns))
```

## Min Max Scaling:

```python
scaler = MinMaxScaler()
scaled = scaler.fit_transform(df)
print(pd.DataFrame(scaled,columns=df.columns))
```

```python
df = pd.DataFrame({
    "Profit":[-100,-50,0,50,100]
})

print(df)
```

## MaxABS Scaler:

```python
scaler = MaxAbsScaler()
scaled = scaler.fit_transform(df)
print(pd.DataFrame(scaled,columns=df.columns))
```

## Robust Scaler:


```python
df = pd.DataFrame({
    "Salary":[30000,35000,40000,45000,500000]
})

print(df)
```

```python
scaler = RobustScaler()
scaled = scaler.fit_transform(df)
print(pd.DataFrame(scaled,columns=df.columns))
```
