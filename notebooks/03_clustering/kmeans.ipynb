### Importing Libraries:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
df = pd.DataFrame({
    "Height": [150, 152, 155, 175, 178, 180],
    "Weight": [45, 46, 48, 72, 75, 78]
})

print(df)
```

# K-Means:

```python
kmeans = KMeans(n_clusters=2,random_state=42)
```

```python
df["cluster"] = kmeans.fit_predict(df[["Height","Weight"]])
```

```python
plt.scatter(
    df["Height"],
    df["Weight"],
    c=df["cluster"],
    s=100
)

plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("K-Means Clustering")

plt.show()
```
